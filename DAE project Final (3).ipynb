```python
# Columns that have missing values
columns_to_fill = [
    'population',
    'population_density',
    'aged_65_older',
    'aged_70_older',
    'gdp_per_capita',
    'cardiovasc_death_rate',
    'diabetes_prevalence',
    'handwashing_facilities',
    'life_expectancy'
]

# Fill missing values using the median value for each country
for col in columns_to_fill:
    df2[col] = df2.groupby('location')[col].transform(
        lambda x: x.fillna(x.median())
    )
```

```python
indices = np.random.permutation(len(X))
```

```python
train_size = int(0.8 * len(X))

print("Training size:", train_size)
print("Testing size:", len(X) - train_size)
```

```python
train_indices = indices[:train_size]
test_indices = indices[train_size:]
```

```python
X_train = X.iloc[train_indices].copy()
X_test = X.iloc[test_indices].copy()
```

```python
print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
```

```python
train_mean = X_train.mean()
train_std = X_train.std()
```

```python
X_train_scaled = (X_train - train_mean) / train_std
```

```python
X_test_scaled = (X_test - train_mean) / train_std
```

```python
print(X_train_scaled.shape)
print(X_test_scaled.shape)
```

```python
print(X_train_scaled.describe().T)
```

```python
print("Training means:")
print(X_train_scaled.mean())
```

```python
print("Training standard deviations:")
print(X_train_scaled.std())
```

```python
print(
    "Training NaN values:",
    X_train_scaled.isnull().sum().sum()
)

print(
    "Testing NaN values:",
    X_test_scaled.isnull().sum().sum()
)
```

```python
print(
    "Training infinite values:",
    np.isinf(X_train_scaled).sum().sum()
)

print(
    "Testing infinite values:",
    np.isinf(X_test_scaled).sum().sum()
)
```

```python
X_train_clean = X_train_scaled.to_numpy()
X_test_clean = X_test_scaled.to_numpy()
```

```python
print("Training array shape:", X_train_clean.shape)
print("Testing array shape:", X_test_clean.shape)
```

```python
noise_factor = 0.1
```

```python
noise_train = noise_factor * np.random.normal(
    0,
    1,
    X_train_clean.shape
)
```

```python
noise_test = noise_factor * np.random.normal(
    0,
    1,
    X_test_clean.shape
)
```

```python
X_train_noisy = X_train_clean + noise_train
X_test_noisy = X_test_clean + noise_test
```

```python
print("========== FINAL PREPROCESSING CHECK ==========")

print("Original dataset:", df.shape)

print("Feature dataset:", X.shape)

print("Training data:", X_train_clean.shape)

print("Testing data:", X_test_clean.shape)

print("Training noisy data:", X_train_noisy.shape)

print("Testing noisy data:", X_test_noisy.shape)

print("Training NaN:",
      np.isnan(X_train_clean).sum())

print("Testing NaN:",
      np.isnan(X_test_clean).sum())

print("Training infinite:",
      np.isinf(X_train_clean).sum())

print("Testing infinite:",
      np.isinf(X_test_clean).sum())

print("==============================================")
```

```python
pair_columns = [
    'total_cases',
    'new_cases',
    'total_deaths',
    'new_deaths',
    'total_cases_per_million'
]

pair_data = df[pair_columns].copy()

# Create a 5 × 5 pair plot
fig, axes = plt.subplots(
    len(pair_columns),
    len(pair_columns),
    figsize=(15, 15)
)

for i, row_col in enumerate(pair_columns):
    for j, col_col in enumerate(pair_columns):

        ax = axes[i, j]

        # Diagonal: histogram
        if i == j:
            ax.hist(
                pair_data[row_col].dropna(),
                bins=20,
                edgecolor='black'
            )

        # Off-diagonal: scatter plot
        else:
            ax.scatter(
                pair_data[col_col],
                pair_data[row_col],
                alpha=0.4
            )

        # X-axis labels only on bottom row
        if i == len(pair_columns) - 1:
            ax.set_xlabel(col_col, rotation=45)

        else:
            ax.set_xticklabels([])

        # Y-axis labels only on first column
        if j == 0:
            ax.set_ylabel(row_col)

        else:
            ax.set_yticklabels([])

plt.suptitle(
    "Pair Plot of Selected COVID-19 Variables",
    fontsize=18,
    y=1.02
)

plt.tight_layout()
plt.show()
```
